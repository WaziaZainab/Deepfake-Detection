```python
def extract_frames(video_path, num_frames=20, img_size=224):
    
    cap = cv2.VideoCapture(video_path)
    
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    
    if total_frames < num_frames:
        frame_indices = np.arange(total_frames)
    else:
        frame_indices = np.linspace(0, total_frames-1, num_frames, dtype=int)
    
    frames = []
    current_frame = 0
    frame_pos = 0
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
            
        if frame_pos < len(frame_indices) and current_frame == frame_indices[frame_pos]:
            
            frame = cv2.resize(frame, (img_size, img_size))
            frame = frame / 255.0
            frames.append(frame)
            frame_pos += 1
        
        current_frame += 1
    
    cap.release()
    
    return np.array(frames)
```

```python
import cv2
import numpy as np

def extract_frames(video_path, num_frames=20, img_size=224):

    cap = cv2.VideoCapture(video_path)

    frames = []

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    if total_frames == 0:
        return np.array([])

    frame_indices = np.linspace(0, total_frames-1, num_frames, dtype=int)

    for idx in frame_indices:
        cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
        ret, frame = cap.read()

        if ret:
            frame = cv2.resize(frame, (img_size, img_size))
            frame = frame / 255.0
            frames.append(frame)

    cap.release()

    return np.array(frames)
```

```python
import numpy as np
from decord import VideoReader, cpu

def extract_frames(video_path, num_frames=20, img_size=224):

    try:
        vr = VideoReader(video_path, ctx=cpu(0))

        total_frames = len(vr)

        if total_frames == 0:
            return np.array([])

        frame_indices = np.linspace(0, total_frames-1, num_frames, dtype=int)

        frames = []

        for idx in frame_indices:
            frame = vr[idx].asnumpy()

            import cv2
            frame = cv2.resize(frame, (img_size, img_size))
            frame = frame / 255.0

            frames.append(frame)

        return np.array(frames)

    except:
        return np.array([])
```
